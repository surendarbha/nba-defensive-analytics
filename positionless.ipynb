```python
from nba_api.stats.endpoints import leaguedashplayershotlocations
import pandas as pd
```

```python
def pull_player_defense_by_zone(s='2024-25',
                                s_type='Regular Season',
                                p_mode='Totals'):

    req = leaguedashplayershotlocations.LeagueDashPlayerShotLocations(
        season=s,
        distance_range="By Zone",
        season_type_all_star=s_type, # season type
        per_mode_detailed=p_mode # not totally sure
    )

    df = req.get_data_frames()[0]
    return df

df = pd.DataFrame(pull_player_defense_by_zone())

df
```

```python
def ave_fg_pct(zone):
    return DF[zone]['FG_PCT'].mean()

zones = ['Restricted Area', 'In The Paint (Non-RA)', 'Mid-Range', 'Above the Break 3', 'Left Corner 3', 'Right Corner 3']

base_line = [ave_fg_pct(zone) for zone in zones]

def fg_pct_zones(row):
    return [row[zone]['FG_PCT'] for zone in zones]
```

```python
from scipy.stats import chisquare
import numpy as np

z_score_stats = {}

def std_fg_pct_zones():
    return np.array([np.std(DF[zone]['FG_PCT']) for zone in zones])

std = std_fg_pct_zones()
ave = np.array(base_line)

def z_score_zones(row):
    rw = np.array(fg_pct_zones(row))
    return (rw - ave) / std

for index, row in DF.iterrows():
    z_score_stats[row[1]] = z_score_zones(row)
```

```python
import matplotlib.pyplot as plt

names = []
simple_averages = []
simple_standard_devs = []

for k, v in z_score_stats.items():
    simple_averages.append(v.mean())
    simple_standard_devs.append(np.std(v))
    names.append(k)

plt.scatter(simple_averages, simple_standard_devs)
plt.xlabel("average")
plt.ylabel("standard dev")
plt.show()
```

```python
players = []

for i in range(len(names)):
    players.append((names[i], simple_averages[i], simple_standard_devs[i]))

players.sort(key=lambda x: x[2])

for p in players:
    if p[1] < -1.5:
        print(p)

z_score_stats["Phillip Wheeler"]
```
